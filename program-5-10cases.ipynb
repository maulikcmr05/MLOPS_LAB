```python
# ============================================
# MACHINE LEARNING PIPELINE
# Iris Dataset + 10 Test Cases
# ============================================

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report


# ============================================
# Step 1: Data Loader Class
# ============================================

class DataLoader:

    def __init__(self):
        self.X, self.y = load_iris(return_X_y=True)

    def split(self, test_size=0.3, random_state=42):
        return train_test_split(
            self.X,
            self.y,
            test_size=test_size,
            random_state=random_state
        )


# ============================================
# Step 2: Preprocessor Class
# ============================================

class Preprocessor:

    def __init__(self):
        self.scaler = StandardScaler()

    def fit_transform(self, X_train):
        return self.scaler.fit_transform(X_train)

    def transform(self, X_test):
        return self.scaler.transform(X_test)


# ============================================
# Step 3: ML Model Class
# ============================================

class MLModel:

    def __init__(self):
        self.model = DecisionTreeClassifier(random_state=42)

    def train(self, X_train, y_train):
        self.model.fit(X_train, y_train)

    def predict(self, X_test):
        return self.model.predict(X_test)


# ============================================
# Step 4: Evaluator Class
# ============================================

class Evaluator:

    def __init__(self, y_true, y_pred):
        self.y_true = y_true
        self.y_pred = y_pred

    def report(self):
        print("\nClassification Report:\n")
        print(classification_report(self.y_true, self.y_pred))


# ============================================
# Step 5: Main ML Application Class
# ============================================

class MLApplication:

    def __init__(self):
        self.loader = DataLoader()
        self.preprocessor = Preprocessor()
        self.model = MLModel()

    def run(self):

        # Load data
        X_train, X_test, y_train, y_test = self.loader.split()

        # Preprocess data
        X_train_scaled = self.preprocessor.fit_transform(X_train)
        X_test_scaled = self.preprocessor.transform(X_test)

        # Train model
        self.model.train(X_train_scaled, y_train)

        # Predict
        y_pred = self.model.predict(X_test_scaled)

        # Evaluate
        evaluator = Evaluator(y_test, y_pred)
        evaluator.report()


# ============================================
# Step 6: Run ML Application
# ============================================

print("===================================")
print("       ML APPLICATION OUTPUT")
print("===================================")

app = MLApplication()
app.run()


# ============================================
# Step 7: 10 Test Cases
# ============================================

def test_case_1():
    loader = DataLoader()
    return len(loader.X) == 150


def test_case_2():
    loader = DataLoader()
    return len(loader.y) == 150


def test_case_3():
    loader = DataLoader()
    return loader.X.shape[1] == 4


def test_case_4():
    loader = DataLoader()
    return loader.X is not None


def test_case_5():
    loader = DataLoader()
    X_train, X_test, y_train, y_test = loader.split()
    return len(X_test) == 45


def test_case_6():
    loader = DataLoader()
    X_train, X_test, y_train, y_test = loader.split()
    return len(X_train) == 105


def test_case_7():
    preprocessor = Preprocessor()
    return preprocessor.scaler is not None


def test_case_8():
    model = MLModel()
    return model.model is not None


def test_case_9():
    loader = DataLoader()

    X_train, X_test, y_train, y_test = loader.split()

    preprocessor = Preprocessor()
    X_train = preprocessor.fit_transform(X_train)

    model = MLModel()
    model.train(X_train, y_train)

    return True


def test_case_10():
    loader = DataLoader()
    return set(loader.y) == {0, 1, 2}


# ============================================
# Step 8: Run 10 Test Cases in Order
# ============================================

print("\n===================================")
print("          10 TEST CASES")
print("===================================\n")

tests = [
    test_case_1,
    test_case_2,
    test_case_3,
    test_case_4,
    test_case_5,
    test_case_6,
    test_case_7,
    test_case_8,
    test_case_9,
    test_case_10
]

passed = 0

for i, test in enumerate(tests, start=1):

    try:
        result = test()

        if result:
            print(f"Test Case {i}: PASS")
            passed += 1
        else:
            print(f"Test Case {i}: FAIL")

    except Exception as e:
        print(f"Test Case {i}: FAIL")
```
